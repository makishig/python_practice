```python
d = {'today' : date.today(),
     'delta' : date(2025, 3, 1) - date.today()}
```

```python
d
```

```python
import pickle  #データを直列化
pickle.dumps(d)
```

```python
with open('date.pkl', 'wb') as f:
    pickle.dump(d, f)
    
with open('date.pkl', 'rb') as f:
    new_d = pickle.load(f)

new_d
```
