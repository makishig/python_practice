```python
import matplotlib.pyplot as plt
import matplotlib.style
matplotlib.style.use('ggplot')

fig, ax = plt.subplots()
ax.legend(loc='best')
ax.bar([1, 2, 3], [5, 8, 2], tick_label=['spam', 'ham', 'egg'])

ax.plot([1, 2, 3], [2, 4, 9], label='Legend label', color='blue')

rng = np.random.default_rng(123)
x = rng.random(50)
y = rng.random(50)
ax.scatter(x, y, color='green')
plt.show()
```

```python
mu = 100
x0 = rng.normal(mu, 20, 100)
x1 = rng.normal(mu, 15, 100)
x2 = rng.normal(mu, 10, 100)
fig, ax = plt.subplots()
labels = [20, 15, 10]
ax.hist((x0, x1, x2), label=labels)
ax.legend()
```

```python
labels = ['spam', 'ham', 'egg']
explode = [0, 0.2, 0]
x = [10, 3, 1]
fig, ax = plt.subplots()
ax.pie(x, labels=labels, startangle=90,
       counterclock=False, shadow=True, autopct='%1.2f%%', explode=explode)
plt.show()
```
