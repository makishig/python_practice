```python
import json
import shutil
from pathlib import Path

#分類ルールの読み込み
with open("categories.json", "r", encoding="utf-8") as f:
    categories = json.load(f)

#整理するフォルダ
folder = Path(r'C:\Users\shige\OneDrive\ドキュメント\研究室\調査許可申請書(練習用4)')

#ファイル整理の関数化
def organize_files(folder, categories):
    print('\n次のように整理します....')
    
    move_list = []
    for file in folder.iterdir():
        
        if file.is_file():
            suffix = file.suffix.lower()
         
            for category, extensions in categories.items():
                
                if suffix in extensions:
                    
                    move_list.append((file, category))
                    #destination_folder = folder/category
                    #destination_folder.mkdir(exist_ok=True)
                    #shutil.move(file, destination_folder/file.name)
                    #print(file.name, '->', category)  #移動先を表示
                    break
            else:
                category = 'その他'
                move_list.append((file, category))
                
                #destination_folder = folder/'その他'
                #destination_folder.mkdir(exist_ok=True)
                #shutil.move(file, destination_folder/file.name)
                #print(file.name, '->', 'その他')
    print('\n===移動予定===')
    if not move_list:
        print('移動するファイルはありません')
        return
    for file, category in move_list:
        print(file.name, '->', category)
    
    choice = input('この内容で整理しますか？(y or n):')
    if choice != 'y':
        print('\n整理をキャンセルしました')
        return
    
    for file, category in move_list:
        destination_folder = folder/category
        destination_folder.mkdir(exist_ok=True)
        destination = destination_folder/file.name
        
        if destination.exists():   #移動先に同じ名前のファイルがあったとき
            print('\n同じ名前のファイルがあるためスキップ', file.name)
            
        else:    
            try:
                shutil.move(file, destination)
                
            except PermissionError:
                print('\n移動できませんでした（ファイルが使用中の可能性があります）', 'file.name')
        
    
    print('\n整理が完了しました')
        
        

#ルール保存の関数
def save_categories(categories):
    with open('categories.json', 'w', encoding='utf-8') as f:
        json.dump(categories, f, ensure_ascii=False, indent=4)
        
#分類ルール設定の関数                    
def manage_categories(categories):
    while True:

        print("\n===== 分類ルール設定 =====")
        print("1. 分類を追加")
        print("2. 分類を変更")
        print("3. 分類を削除")
        print("4. 分類を確認")    
        print('5. 戻る')


        choice = input("番号を選択してください：")

        if choice == "1":   #分類を追加

            category = input("追加する分類名：")
            extensions = input("拡張子をカンマ区切りで入力：").split(',')
            categories[category] = extensions
            save_categories(categories) #jsonファイルに書き込み

            print(f"\n{category}を追加しました")
            
        elif choice == "2":  #分類を変更

            print("\n現在の分類")

            for category, extensions in categories.items():
                print(category, "→", extensions)

            category = input("変更する分類名：")

            if category in categories:

                new_extensions = input("新しい拡張子をカンマ区切りで入力：").split(',')
                categories[category] = new_extensions
                save_categories(categories) #jsonファイルに書き込み

                print(f"\n{category}を変更しました")            
            
            else:
                print('\nその分類は存在しません')
                
        elif choice == '3':  #分類を削除
            print('\n分類を削除します')
        
            print('\n現在の分類')
            for category, extensions in categories.items():
                print(category, '->', extensions)
        
            category = input('削除する分類名：')
            if category in categories:
                del categories[category]
                save_categories(categories) #jsonファイルに書き込み
                print(f'\n[{category}]を削除しました')
            else:
                print('\nその分類は存在しません')            
        
        elif choice == '4':  #分類を確認
            print('\n現在の分類を確認')    
            for category, extension in categories.items():
                print(category, '->', extension)
                
        elif choice == '5':  #戻る
            break
        
        else:
            print('\n1~5を入力してください')
            
            
            
            
            
            
            
            
            
#メインメニュー
while True:
    
    print('\n===ファイル整理ツール===')
    print('1. 分類ルール設定')
    print('2. ファイルを整理')
    print('3. 終了')
    
    choice = input('番号を入力してください: ')
    
    if choice == '1':
        manage_categories(categories)
        
    elif choice == '2':
        organize_files(folder, categories)
        
    elif choice == '3':
        print('\n終了します')
        break
    
    else:
        print('\n1~3を入力してください')
```

```python
#実行ボタンでファイルを移動　エラーをGUI上で表示
import json
from pathlib import Path
import tkinter as tk
from tkinter import filedialog, messagebox
import shutil

with open("categories.json", "r", encoding="utf-8") as f:
    categories = json.load(f)
    
def organize_files(folder, categories): 
        
    move_list = []
    for file in folder.iterdir():
        
        if file.is_file():
            suffix = file.suffix.lower()
         
            for category, extensions in categories.items():
                
                if suffix in extensions:
                    
                    move_list.append((file, category))
                    #destination_folder = folder/category
                    #destination_folder.mkdir(exist_ok=True)
                    #shutil.move(file, destination_folder/file.name)
                    #print(file.name, '->', category)  #移動先を表示
                    break
            else:
                category = 'その他'
                move_list.append((file, category))
                
    return move_list

def select_folder():
    folder = filedialog.askdirectory()
    folder_var.set(folder)

def organize_selected_folder():
    
    global move_list  #関数内で関数外の move_list を書き換えるため、スコープを変更
    
    if not folder_var.get():   
        messagebox.showwarning('警告', '先にフォルダを選択してください')
        return
    
    folder = Path(folder_var.get())
    move_list = organize_files(folder, categories)
    
    text_box.delete('1.0', tk.END) 
    
    if not move_list:
        text_box.insert(tk.END, 'ファイルが存在しません')
        return
    
    text_box.insert(tk.END, '===移動予定===\n\n')
    for file, category in move_list:
        text_box.insert(tk.END, f'{file.name} -> {category}\n')
    
def execute_move():
    if not move_list:
        messagebox.showinfo('確認', '移動するファイルがありません')
        return
    
    answer = messagebox.askyesno('確認', '表示されている内容でファイルを移動しますか？')
    
    if not answer:   #はい　なら　True　いいえ　なら　False がanswerに入る
        return
        
    for file, category in move_list:
        destination_folder = file.parent/category  #file.parent　ファイルが入っている親フォルダ
        destination_folder.mkdir(exist_ok=True)
        
        destination = destination_folder/file.name
        if destination.exists():
            text_box.insert(tk.END, '同じ名前のファイルがあるためスキップ:', file.name, '\n')
        else:
            try:
                shutil.move(file, destination)
            except PermissionError:
                text_box.insert(tk.END, '移動できませんでした;', file.name, '\n')
    messagebox.showinfo('完了', 'ファイルの整理が完了しました')
        

root = tk.Tk()

move_list =[]  #実際に移動させるためにGUI全体でmove_listを使えるようにしなければいけない

root.title('ファイル整理ツール')
root.geometry('500x300')

folder_var = tk.StringVar()

label = tk.Label(root, textvariable=folder_var) 
label.pack()

select_button = tk.Button(root, text='フォルダを選択', command=select_folder)
select_button.pack()

text_box = tk.Text(root, width=70, height=15)
text_box.pack()

organize_button = tk.Button(root, text='移動予定を確認', command=organize_selected_folder)
organize_button.pack()

execute_button = tk.Button(root, text='実行', command=execute_move) #実行ボタン
execute_button.pack()

root.mainloop()
```

```python
#move_list をより安全に　いつ書き換えられたのかを明確に　
#move_list を作成した後に、フォルダの中身が変わった場合
#実行後にmove_listを空にする
import json
from pathlib import Path
import tkinter as tk
from tkinter import filedialog, messagebox
import shutil

with open("categories.json", "r", encoding="utf-8") as f:
    categories = json.load(f)
    
def organize_files(folder, categories): 
        
    move_list = []
    for file in folder.iterdir():
        
        if file.is_file():
            suffix = file.suffix.lower()
         
            for category, extensions in categories.items():
                
                if suffix in extensions:
                    
                    move_list.append((file, category))
                    
                    break
            else:
                category = 'その他'
                move_list.append((file, category))
                
    return move_list

def select_folder():
    folder = filedialog.askdirectory()
    
    if not folder:   #フォルダ選択画面でキャンセルした場合
        return
    
    folder_var.set(folder)
    move_list.clear()      #フォルダを変更した際にmove_listが前のフォルダの情報を残さないように
    #global move_list はいらない
    #gloabal　は変数を再代入する際に必要　今回は.clear()で変更するだけで書き換えていないため必要ない
    text_box.delete('1.0', tk.END)    #前のフォルダの移動予定を画面に残さないように

def organize_selected_folder():
    
    global move_list  #関数内で関数外の move_list を再代入し、書き換えるため、スコープを変更
    
    if not folder_var.get():   
        messagebox.showwarning('警告', '先にフォルダを選択してください')
        return
    
    folder = Path(folder_var.get())
    move_list = organize_files(folder, categories)
    
    text_box.delete('1.0', tk.END) 
    
    if not move_list:
        text_box.insert(tk.END, 'ファイルが存在しません')
        return
    
    text_box.insert(tk.END, '===移動予定===\n\n')
    for file, category in move_list:
        text_box.insert(tk.END, f'{file.name} -> {category}\n')
    
def execute_move():
    if not move_list:
        messagebox.showinfo('確認', '移動するファイルがありません')
        return
    
    answer = messagebox.askyesno('確認', '表示されている内容でファイルを移動しますか？')
    
    if not answer:   #はい　なら　True　いいえ　なら　False がanswerに入る
        return
        
    for file, category in move_list:
        destination_folder = file.parent/category  #file.parent　ファイルが入っている親フォルダ
        
        if not file.exists(): #実行時にファイルがまだ存在するかを確認　そのためにfor内に設置
            text_box.insert(tk.END, f'ファイルが見つからないためスキップ:{file.name}\n')
            continue  #次のmove_listの要素へ
        
        destination_folder.mkdir(exist_ok=True)
        
        destination = destination_folder/file.name
        if destination.exists():
            text_box.insert(tk.END, '同じ名前のファイルがあるためスキップ:', file.name, '\n')
        else:
            try:
                shutil.move(file, destination)
            except PermissionError:
                text_box.insert(tk.END, '移動できませんでした;', file.name, '\n')
    
    move_list.clear()  #実行後にリストを空にする
    messagebox.showinfo('完了', 'ファイルの整理が完了しました')
        

root = tk.Tk()

move_list =[]

root.title('ファイル整理ツール')
root.geometry('500x300')

folder_var = tk.StringVar()

label = tk.Label(root, textvariable=folder_var) 
label.pack()

select_button = tk.Button(root, text='フォルダを選択', command=select_folder)
select_button.pack()

text_box = tk.Text(root, width=70, height=15)
text_box.pack()

organize_button = tk.Button(root, text='移動予定を確認', command=organize_selected_folder)
organize_button.pack()

execute_button = tk.Button(root, text='実行', command=execute_move) #実行ボタン
execute_button.pack()

root.mainloop()
```
