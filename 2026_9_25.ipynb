```python
from pathlib import Path
#調査許可申請書のディレクトリのパスを取得
folder = Path(r'C:\Users\shige\OneDrive\ドキュメント\研究室\調査許可申請書')

for file in folder.iterdir():
    if file.is_file():
        if file.suffix.lower() == '.pdf':
            print(file.name, '-> PDF')        
        elif file.suffix.lower() == '.pptx':
            print(file.name, '-> パワポ')
        elif file.suffix.lower() == '.docx':
            print(file.name, '-> ward')
        elif file.suffix.lower() == '.xlsx':
            print(file.name, '-> エクセル')
```

```python
folder = Path(r'C:\Users\shige\OneDrive\ドキュメント\研究室\調査許可申請書')

pdf_folder = folder / 'PDF'  #pathは / でくっつけることができる
pdf_folder.mkdir(exist_ok=True)  #.mkdirでディレクトリを作成　存在していたら作らなくてもよいと指定
```

```python
import shutil
folder = Path(r'C:\Users\shige\OneDrive\ドキュメント\研究室\調査許可申請書')

pdf_folder = folder / 'PDF'
ward_folder = folder / 'Ward' 
exel_folder = folder / 'Exel'
pptx_folder = folder / 'パワポ'

pdf_folder.mkdir(exist_ok=True)
ward_folder.mkdir(exist_ok=True)
exel_folder.mkdir(exist_ok=True)
pptx_folder.mkdir(exist_ok=True)

for file in folder.iterdir():
    if file.is_file():
        
        if file.suffix.lower() == '.pdf':
            shutil.move(file, file / pdf_folder / file.name)
            
        elif file.suffix.lower() == '.docx':
            shutil.move(file, ward_folder / file.name)
            
        elif file.suffix.lower() == '.xlsx':
            shutil.move(file, exel_folder / file.name)
            
        elif file.suffix.lower() == '.pptx':
            shutil.move(file, pptx_folder / file.name)
        
```

```python
folder = Path(r'C:\Users\shige\OneDrive\ドキュメント\研究室\調査許可申請書(練習用)')

for file in folder.iterdir():
    print(file)  #file　はパス
```

```python
categories = {
    ".pdf": "PDF",
    ".jpg": "画像",
    ".jpeg": "画像",
    ".png": "画像",
    ".csv": "CSV",
    ".xlsx": "Excel",
    ".docx": "Word",
    ".pptx": "PowerPoint"
}
folder = Path(r'C:\Users\shige\OneDrive\ドキュメント\研究室\調査許可申請書(練習用)')

for file in folder.iterdir():
    if file.is_file():
        
        suffix = file.suffix.lower()
        
        if suffix in categories:
            category = categories[suffix]
            
            destination_folder = folder / category
            destination_folder.mkdir(exist_ok=True)
            
            shutil.move(file, destination_folder/file.name)
```
